## 1. Wstęp: Historia, rozwój przez PEP oraz cechy języka Python

### 1.1. Historia języka Python i organizacja jego rozwoju (PEP)

Twórcą języka Python jest holenderski programista **Guido van Rossum**. Sama nazwa języka nie pochodzi od gada, lecz od ulubionego serialu komediowego BBC twórcy — *„Latający Cyrk Monty Pythona”* (*Monty Python’s Flying Circus*), dlatego w oficjalnej dokumentacji często można natknąć się na żartobliwe nawiązania do skeczy tej grupy (np. zmienne `spam` oraz `eggs`).

Prace nad pierwszym interpreterem Pythona rozpoczęły się w grudniu **1989 roku** w **CWI (*Centrum Wiskunde & Informatica* — Centrum Matematyki i Informatyki)** w Amsterdamie. Python miał być bezpośrednim następcą współtworzonego tam wcześniej edukacyjnego języka **ABC** — zachowującym jego wyjątkową czytelność i prostotę składni, ale otwartym na komunikację z systemem operacyjnym (podobnie jak język C i powłoka Unix).

**Najważniejsze kamienie milowe rozwoju języka:**
- **1991 (Python 0.9.0) – 1994 (Python 1.0):** powstają fundamenty języka — wcięcia wyznaczające bloki kodu, podstawowe typy danych (`str`, `list`, `dict`), moduły, wyjątki oraz funkcje.
- **2000 (Python 2.0) oraz 2001 (powstanie PSF):** od wersji 2.1 język jest rozwijany jako projekt *Open Source* pod opieką organizacji non-profit **Python Software Foundation (PSF)** ([https://www.python.org/psf-landing/](https://www.python.org/psf-landing/)). Wprowadzony zostaje m.in. mechanizm *list comprehensions* oraz odśmiecanie pamięci wykrywające cykle referencji.
- **Rola BDFL → Python Steering Council:** przez prawie trzy dekady Guido van Rossum pełnił w społeczności honorową funkcję **BDFL (*Benevolent Dictator for Life* — Życzliwy Dyktator na Życie)**, podejmując ostateczne decyzje w spornych kwestiach rozwoju języka. W lipcu 2018 roku (po burzliwych dyskusjach wokół dokumentu PEP 572 wprowadzającego operator przypisania `:=`) Guido zrezygnował z roli BDFL. Od 2019 roku nadzór nad rozwojem języka sprawuje wybierana co roku pięcioosobowa rada **Python Steering Council** ([PEP 13](https://peps.python.org/pep-0013/)). Sam twórca pracował m.in. w Google (2005–2012), Dropbox (2013–2019) oraz Microsoft (od 2020 r., wspierając projekt przyspieszenia interpretera CPython).
- **2008 (Python 3.0) → 2020 (koniec Pythona 2):** gałąź Python 3 została zaprojektowana tak, aby naprawić wczesne błędy projektowe języka (m.in. ujednolicić obsługę znaków Unicode w typie `str` oraz uczynić `print()` zwykłą funkcją). Przez lata obie gałęzie współistniały, aż **1 stycznia 2020 r.** wsparcie dla Pythona 2.7 zostało definitywnie zakończone.
- **Python 3.11 – 3.14 (współczesna era wydajności i ergonomii):** w najnowszych wydaniach (w tym **Python 3.14**, na którym pracujemy) wprowadzono bardzo czytelne, precyzyjne komunikaty błędów wskazujące dokładne miejsce pomyłki w linii, znacząco przyspieszono wykonanie kodu w CPythonie oraz rozwinięto eksperymentalny tryb wielowątkowości bez globalnej blokady interpretera (*free-threaded Python*, [PEP 703](https://peps.python.org/pep-0703/)).

### Jak rozwija się Python? Dokumenty PEP (*Python Enhancement Proposals*)

Każda istotna zmiana w składni języka, bibliotece standardowej lub procesach organizacyjnych musi zostać opisana w formalnym dokumencie technicznym nazywanym **PEP (*Python Enhancement Proposal*)** — [https://peps.python.org/](https://peps.python.org/).

Zgodnie z dokumentem **[PEP 1](https://peps.python.org/pep-0001/)**, każdy PEP zawiera:
1. Zwięzły opis proponowanego rozwiązania i jego specyfikację techniczną,
2. Motywację (*Rationale*) — dlaczego zmiana jest potrzebna,
3. Analizę wstecznej kompatybilności oraz odrzuconych alternatyw,
4. Aktualny status dokumentu (*Draft*, *Accepted*, *Final*, *Rejected*, *Superseded*).

Trzy dokumenty PEP, które zna każdy programista Pythona:
- **[PEP 1](https://peps.python.org/pep-0001/)** — opisuje sam system dokumentów PEP,

### 1.2. Wybrane cechy techniczne języka Python

Python jest językiem **ogólnego przeznaczenia** (*general-purpose language*), którego nadrzędną ideą jest czytelność i klarowność kodu źródłowego. Standardową implementacją języka jest **CPython** (napisany w C), ale istnieją również inne implementacje specjalistyczne:
- **PyPy** — implementacja z kompilatorem JIT (*Just-In-Time*), przyspieszająca długotrwałe pętle obliczeniowe w czystym Pythonie,
- **Jython** — uruchamiana na Wirtualnej Maszynie Javy (JVM),
- **IronPython** — zintegrowana z platformą .NET,
- **MicroPython / CircuitPython** — odchudzona wersja na mikrokontrolery.

Python nie narzuca jednego paradygmatu — pozwala swobodnie łączyć **programowanie strukturalne**, **obiektowe** oraz **funkcyjne**.

Przeanalizujmy najważniejsze cechy techniczne Pythona na pojedynczych, krótkich przykładach w komórkach poniżej.

---

#### Cecha 1: Typowanie dynamiczne (*Dynamic typing*)
W przeciwieństwie do języków typowanych statycznie (takich jak C, C++, Java czy C#), w Pythonie **typ jest powiązany z obiektem w pamięci, a nie z samą nazwą zmiennej**. Ta sama zmienna może w trakcie działania programu wskazywać najpierw na liczbę całkowitą, a chwilę później na tekst.

```python
zmienna = 42
print("Wartość:", zmienna, "| Typ:", type(zmienna))
```

```python
# Tej samej zmiennej przypisujemy teraz obiekt typu str (napis):
zmienna = "czterdzieści dwa"
print("Wartość:", zmienna, "| Typ:", type(zmienna))
```

#### Cecha 2: Typowanie silne (*Strong typing*)
Dynamiczne typowanie **nie oznacza** słabego typowania! W Pythonie nie dochodzi do ukrytego, przypadkowego sklejania tekstu z liczbą (jak np. w języku JavaScript, gdzie `"5" + 3` daje `"53"`). Próba dodania napisu i liczby zakończy się czytelnym wyjątkiem `TypeError` — konwersję należy wykonać jawnie (*Explicit is better than implicit*).

```python
# Jawna konwersja (rzutowanie) tekstu na liczbę całkowitą przed dodawaniem:
tekst_liczba = "5"
wynik = int(tekst_liczba) + 3
print("Poprawny wynik po jawnej konwersji:", wynik)
```

#### Cecha 3: Wszystko jest obiektem i przekazywane jest przez referencję (`id()`)
W Pythonie każda wartość (nawet prosta liczba `100` czy funkcja) jest obiektem umieszczonym pod pewnym adresem w pamięci, a zmienna jest jedynie **etykietą (referencją)** wskazującą na ten obiekt. Unikalny identyfikator obiektu w pamięci sprawdzamy wbudowaną funkcją `id()` ([https://docs.python.org/3.14/library/functions.html#id](https://docs.python.org/3.14/library/functions.html#id)). Niepotrzebne już obiekty są automatycznie usuwane z pamięci przez **odśmiecacz pamięci (*Garbage Collector*)**.

```python
lista_a = [1, 2, 3]
lista_b = lista_a  # lista_b wskazuje na TEN SAM obiekt w pamięci co lista_a!

print("Identyfikator lista_a:", id(lista_a))
print("Identyfikator lista_b:", id(lista_b))
print("Czy to ten sam obiekt w pamięci?", lista_a is lista_b)
```

#### Cecha 4: Brak słów kluczowych `private`/`protected` — konwencja podkreślenia
W Pythonie nie ma sztywnych modyfikatorów dostępu `private` znanych z C++ czy Javy. Zamiast tego stosuje się umowę społeczną („wszyscy jesteśmy dorosłymi programistami”) opartą na pojedynczym (`_zmienna`) lub podwójnym (`__zmienna`) podkreśleniu oraz mechanizmie właściwości (`@property`), który szczegółowo poznamy w Module 7 (OOP).

#### Cecha 5: Funkcje z argumentami domyślnymi oraz zmienną liczbą argumentów
Python pozwala niezwykle elastycznie definiować funkcje przyjmujące wartości domyślne oraz dowolną liczbę argumentów pozycyjnych (`*args`) i nazwanych (`**kwargs`):

```python
def powitanie(imie, uczelnia="UWM w Olsztynie"):
    return f"Cześć {imie}, witamy na {uczelnia}!"


print(powitanie("Anna"))
print(powitanie("Piotr", uczelnia="Politechnika"))
```

### 2.1. Wcięcia (*Indentation*) oraz łamanie długich linii

> **Sekcja w PEP 8:** [https://peps.python.org/pep-0008/#indentation](https://peps.python.org/pep-0008/#indentation) oraz [https://peps.python.org/pep-0008/#maximum-line-length](https://peps.python.org/pep-0008/#maximum-line-length)

#### Wprowadzenie
W kodzie języka Python nie używamy nawiasów klamrowych `{ ... }` znanych z C++, Javy czy C# do wyznaczania bloków kodu (ciała warunku, pętli, funkcji czy klasy). W Pythonie **o przynależności instrukcji do bloku decyduje wcięcie**:
- Każdy poziom zagnieżdżenia to dokładnie **4 spacje**.
- W Pythonie 3 **mieszanie tabulatorów i spacji jako wcięć w tym samym bloku jest zabronione** i powoduje błąd składniowy `TabError` (edytor VS Code automatycznie zamienia naciśnięcie klawisza `Tab` na 4 spacje).
- Zalecana maksymalna długość pojedynczej linii kodu według PEP 8 wynosi **79 znaków** (dla komentarzy i docstringów **72 znaki**; we współczesnych zespołach często dopuszcza się limit 88 lub 100 znaków, o ile jest stosowany konsekwentnie).

Zobaczmy na pięciu osobnych przykładach, jak poprawnie stosować wcięcia i łamać długie linie.

#### Przykład 2.1 — Pojedyncze wcięcie bloku kodu (4 spacje)

```python
wynik_punktowy = 120

if wynik_punktowy >= 100:
    print("Zwycięstwo!")
```

#### Przykład 2.2 — Łamanie długiego wywołania funkcji (wyrównanie do nawiasu otwierającego)
Gdy wywołanie funkcji ma wiele argumentów i przekracza długość linii, kolejne wiersze wyrównujemy do znaku otwarcia nawiasu `(`:

```python
def wyslij_wiadomosc(odbiorca, temat, tresc):
    return f"Wysłano do {odbiorca}: [{temat}] {tresc}"


# Wyrównanie drugiego wiersza dokładnie pod pierwszym argumentem po nawiasie '('
wyslane = wyslij_wiadomosc("student@uwm.edu.pl", "Wyniki zaliczenia",
                           "Gratulacje, projekt został oceniony na 5.0!")
print(wyslane)
```

#### Przykład 2.3 — Alternatywne łamanie wywołania funkcji (wiszące wcięcie 4 spacji)
Możemy również przenieść wszystkie argumenty do nowej linii z pojedynczym wcięciem (4 spacje) i zamknąć nawias w osobnej linii:

```python
wyslane = wyslij_wiadomosc(
    "student@uwm.edu.pl",
    "Przypomnienie o zjeździe",
    "Zajęcia zaczynają się o godzinie 9:00.",
)
print(wyslane)
```

#### Przykład 2.4 — Czytelna deklaracja wielowierszowych kolekcji (list, słowników)
Zwróć uwagę na przecinek po ostatnim elemencie (*trailing comma*) — ułatwia on późniejsze dopisywanie kolejnych wierszy i minimalizuje zmiany w systemie Git:

```python
macierz_wartosci = [
    1, 2, 3,
    4, 5, 6,
    7, 8, 9,
]
print(macierz_wartosci)
```

#### Przykład 2.5 — Łamanie długich wyrażeń matematycznych przed operatorem binarnym
> **Dokumentacja PEP 8:** [https://peps.python.org/pep-0008/#should-a-line-break-before-or-after-a-binary-operator](https://peps.python.org/pep-0008/#should-a-line-break-before-or-after-a-binary-operator)

Gdy wyrażenie arytmetyczne lub logiczne nie mieści się w jednej linii, zamykamy je w nawiasach okrągłych `(...)` i **przenosimy operator (`+`, `-`, `*`) na początek nowej linii**. Dzięki temu wzrok od razu widzi w jednej kolumnie, które składniki są dodawane, a które odejmowane:

```python
przychod_brutto = 15_000
koszty_stale = 4_200
koszty_zmienne = 1_800
podatek_dochodowy = 1_710

zysk_netto = (
    przychod_brutto
    - koszty_stale
    - koszty_zmienne
    - podatek_dochodowy
)
print("Zysk netto:", zysk_netto)
```

### 2.2. Puste linie (*Blank Lines*)

> **Sekcja w PEP 8:** [https://peps.python.org/pep-0008/#blank-lines](https://peps.python.org/pep-0008/#blank-lines)

#### Wprowadzenie
Puste linie pełnią w kodzie rolę akapitów:
1. **Dwie puste linie** oddzielają od reszty kodu definicje funkcji najwyższego poziomu (*top-level functions*) oraz definicje klas (`class`).
2. **Jedna pusta linia** oddziela od siebie metody wewnątrz klasy oraz funkcje lokalne (zagnieżdżone wewnątrz innej funkcji).
3. Pojedyncze puste linie stosujemy również oszczędnie wewnątrz funkcji, aby oddzielić jej logiczne etapy (np. przygotowanie danych → obliczenia → zwrócenie wyniku).

#### Przykład 2.6 — Dwie puste linie pomiędzy funkcjami najwyższego rzędu

```python
def pobierz_dane():
    return [10, 20, 30]


def oblicz_sume(elementy):
    return sum(elementy)


print(oblicz_sume(pobierz_dane()))
```

#### Przykład 2.7 — Jedna pusta linia pomiędzy metodami wewnątrz klasy (i dwie po zakończeniu klasy)

```python
class Osoba:

    def __init__(self, imie, nazwisko):
        self.imie = imie
        self.nazwisko = nazwisko

    def przedstaw_sie(self):
        return f"Nazywam się {self.imie} {self.nazwisko}."

    def inicjaly(self):
        return f"{self.imie[0]}.{self.nazwisko[0]}."


jan = Osoba("Jan", "Kowalski")
print(jan.przedstaw_sie(), "| Inicjały:", jan.inicjaly())
```

### 2.3. Organizacja importów (*Imports*)

> **Sekcja w PEP 8:** [https://peps.python.org/pep-0008/#imports](https://peps.python.org/pep-0008/#imports)

#### Wprowadzenie
Zasady umieszczania instrukcji `import`:
1. Importy zawsze umieszczamy **na samym początku pliku** (tuż po opcjonalnym komentarzu opisującym moduł / docstringu).
2. Każdy import zapisujemy w **oddzielnej linii** (`import os` i w nowej linii `import sys`, a **nie** `import os, sys`). Wyjątkiem jest pobieranie kilku elementów z tego samego modułu: `from pathlib import Path, PurePath`.
3. Importy dzielimy na **3 grupy oddzielone od siebie jedną pustą linią**, w ścisłej kolejności:
   - **Grupa 1:** moduły biblioteki standardowej Pythona (np. `import sys`, `from pathlib import Path`),
   - **Grupa 2:** zewnętrzne biblioteki zainstalowane z PyPI (np. `import pandas as pd`, `import seaborn as sns`),
   - **Grupa 3:** własne moduły lokalne należące do naszego projektu.
4. Unikamy zapisu `from modul import *` (*wildcard import*), ponieważ zaśmieca przestrzeń nazw i utrudnia ustalenie, skąd pochodzi dana funkcja.

### 2.4. Konwencje nazewnicze (*Naming Conventions*)

> **Sekcja w PEP 8:** [https://peps.python.org/pep-0008/#naming-conventions](https://peps.python.org/pep-0008/#naming-conventions)  
> **Dodatkowe zestawienie:** [https://realpython.com/python-pep8/#naming-styles](https://realpython.com/python-pep8/#naming-styles)

#### Wprowadzenie
Konsekwentne nazewnictwo pozwala na pierwszy rzut oka rozpoznać, czy dany identyfikator w kodzie jest zmienną, funkcją, klasą czy stałą:

| Element języka | Styl nazewnictwa wg PEP 8 | Przykłady poprawne | Przykłady błędne |
| :--- | :--- | :--- | :--- |
| **Zmienne** | `snake_case` (małe litery, słowa rozdzielone `_`) | `liczba_studentow`, `cena_brutto`, `is_active` | `LiczbaStudentow`, `cenaBrutto`, `ls` |
| **Funkcje i metody** | `snake_case` (najczęściej czasownik określający czynność) | `oblicz_srednia()`, `pobierz_dane()` | `ObliczSrednia()`, `pobierzDane()` |
| **Moduły i pakiety** | krótkie nazwy `snake_case` (bez spacji i myślników!) | `narzedzia_ds.py`, `analiza_danych` | `Narzedzia-DS.py`, `analiza danych.py` |
| **Klasy** | `PascalCase` / `CapWords` (każde słowo wielką literą, bez `_`) | `KlientSklepu`, `RachunekBankowy` | `klient_sklepu`, `rachunek_Bankowy` |
| **Stałe globalne** | `UPPER_CASE_WITH_UNDERSCORES` (wielkie litery z `_`) | `STAWKA_VAT = 0.23`, `MAX_LICZBA_PROB = 3` | `stawka_vat = 0.23`, `MaxLiczbaProb` |

**Dobre praktyki doboru nazw:**
- Unikaj jednoliterowych nazw (`x`, `a`, `l`, `O`, `I`), z wyjątkiem prostych liczników pętli (`i`, `j`) lub wzorów matematycznych. Litery `l` (małe L), `O` (wielkie o) oraz `I` (wielkie i) są w wielu czcionkach niemal nieodróżnialne od cyfr `1` oraz `0`!
- Mimo dynamicznego typowania warto w nazwie zmiennej sugerować jej charakter:
  - dla zmiennych logicznych (`bool`): prefiksy `czy_...` / `is_...` / `has_...` (np. `czy_aktywny`, `is_valid`),
  - dla kolekcji: liczba mnoga lub określenie zbioru (np. `miasta`, `lista_produktow`),
  - dla zliczeń: `liczba_zamowien`, `word_count`.

#### Przykład 2.9 — Czytelne nazewnictwo stałych, klas, funkcji i zmiennych w praktyce

```python
DOMYSLNA_STAWKA_VAT = 0.23


class KoszykZakupowy:

    def __init__(self, ceny_netto):
        self.ceny_netto = ceny_netto

    def oblicz_wartosc_brutto(self):
        suma_netto = sum(self.ceny_netto)
        return round(suma_netto * (1 + DOMYSLNA_STAWKA_VAT), 2)


koszyk_klienta = KoszykZakupowy([100.0, 50.0])
czy_darmowa_dostawa = koszyk_klienta.oblicz_wartosc_brutto() >= 150.0
print("Brutto:", koszyk_klienta.oblicz_wartosc_brutto(), "| Darmowa dostawa:", czy_darmowa_dostawa)
```

### 2.5. Cudzysłowy, apostrofy oraz białe znaki w wyrażeniach

> **Sekcja w PEP 8:** [https://peps.python.org/pep-0008/#string-quotes](https://peps.python.org/pep-0008/#string-quotes) oraz [https://peps.python.org/pep-0008/#whitespace-in-expressions-and-statements](https://peps.python.org/pep-0008/#whitespace-in-expressions-and-statements)

#### Wprowadzenie
1. **Apostrofy (`'...'`) vs cudzysłowy (`"..."`):** w Pythonie oba zapisy tworzą identyczny obiekt typu `str`. PEP 8 zaleca wybranie jednej konwencji i trzymanie się jej konsekwentnie. Jeśli jednak sam napis zawiera w środku cudzysłów, otaczamy go apostrofem (lub odwrotnie), aby uniknąć ukośników ucieczki `\"`. Dla docstringów zawsze używamy potrójnych cudzysłowów `"""..."""`.
2. **Białe znaki (spacje) wewnątrz nawiasów:** **nie** wstawiamy spacji bezpośrednio po nawiasach otwierających `(`, `[`, `{` ani przed zamykającymi `)`, `]`, `}`, a także przed przecinkiem `,` i dwukropkiem `:`.
3. **Odstępy wokół operatorów:** operatory przypisania (`=`, `+=`), porównania (`==`, `!=`, `<`, `>`) oraz arytmetyczne otaczamy pojedynczą spacją z obu stron.
4. **Wyjątek dla argumentów funkcji:** przy określaniu wartości domyślnej lub przekazywaniu argumentu nazwanego w wywołaniu funkcji **nie stawiamy spacji wokół znaku `=`** (np. `round(3.1415, ndigits=2)`).

Zobaczmy każdy z tych przypadków w osobnych komórkach poniżej.
