## 1. Architektura notatnika: Komórki (Cells) oraz Jądro obliczeniowe (Kernel)

### Wprowadzenie
Plik z rozszerzeniem `.ipynb` (*IPython Notebook* / *Jupyter Notebook*) składa się z dwóch współpracujących elementów:
1. **Dokumentu widocznego na ekranie**, podzielonego na niezależne bloki nazywane **komórkami (*cells*)**:
   - **Komórki kodu (`Code`)** — zawierają kod wykonywalny w języku Python,
   - **Komórki tekstowe (`Markdown`)** — zawierają sformatowany opis, nagłówki, wzory matematyczne, tabele oraz wnioski z analizy.
2. **Jądra obliczeniowego (*Kernel*)** — działającego w tle procesu interpretera Pythona (w naszym przypadku ze środowiska `.venv` z Pythonem 3.14), który wykonuje kod przesyłany z komórek i przechowuje w pamięci RAM utworzone zmienne, funkcje oraz wczytane zbiory danych.

**Krok obowiązkowy przed rozpoczęciem pracy:**  
Spójrz w prawy górny róg okna notatnika w VS Code. Powinien być tam widoczny napis **`.venv (Python 3.14.x)`**. Jeśli widzisz napis *Select Kernel*, kliknij go → wybierz *Python Environments...* → wskaż środowisko `.venv` z katalogu naszego projektu.

## 2. Tryby pracy w notatniku i najważniejsze skróty klawiszowe

### Wprowadzenie
Efektywna praca z notatnikiem Jupyter opiera się na rozróżnieniu dwóch trybów, w jakich może znajdować się aktualnie zaznaczona komórka:

1. **Tryb edycji (*Edit Mode*)** — aktywujesz go klawiszem **`Enter`** lub klikając myszą do wnętrza komórki. Wewnątrz komórki miga kursor tekstowy, a wpisywane znaki trafiają do kodu lub tekstu.
2. **Tryb poleceń (*Command Mode*)** — aktywujesz go klawiszem **`Esc`** (lub klikając na margines obok komórki). Kursor wewnątrz komórki znika, a pojedyncze klawisze literowe służą do zarządzania strukturą notatnika.

| Skrót klawiszowy | Tryb | Działanie |
| :--- | :--- | :--- |
| `Shift + Enter` | Oba | Uruchom bieżącą komórkę i przejdź do komórki poniżej |
| `Ctrl + Enter` *(macOS: `Cmd + Enter`)* | Oba | Uruchom bieżącą komórkę i pozostań na niej |
| `Alt + Enter` *(macOS: `Option + Enter`)* | Oba | Uruchom bieżącą komórkę i wstaw nową pustą komórkę poniżej |
| `Esc` | Edit → Command | Wyjdź z edycji tekstu do trybu poleceń notatnika |
| `Enter` | Command → Edit | Wejdź do edycji treści zaznaczonej komórki |
| `A` *(Above)* | Command | Wstaw nową komórkę **powyżej** bieżącej |
| `B` *(Below)* | Command | Wstaw nową komórkę **poniżej** bieżącej |
| `M` | Command | Zmień typ bieżącej komórki na tekstową (**Markdown**) |
| `Y` | Command | Zmień typ bieżącej komórki na kodową (**Code**) |
| `D`, `D` *(dwa razy szybko)* | Command | Usuń zaznaczoną komórkę |

## 3. Współdzielony stan pamięci kernela i pułapki kolejności uruchamiania

### Wprowadzenie
Wszystkie komórki kodu w jednym notatniku korzystają z **tej samej pamięci operacyjnej uruchomionego kernela**. Oznacza to dwie niezwykle ważne rzeczy:
1. Zmienna zdefiniowana w jednej komórce jest od tej pory dostępna we wszystkich innych komórkach (nawet położonych wyżej w dokumencie!).
2. O aktualnej wartości zmiennej decyduje **kolejność, w jakiej uruchamiałeś komórki** (widoczna jako rosnące liczby `[1]`, `[2]`, `[3]` obok komórek), a **nie** ich kolejność wizualna od góry do dołu.

Zobaczmy to krok po kroku na czterech krótkich komórkach.

### Przykład 3.1 — Utworzenie zmiennej w pamięci kernela
Uruchom poniższą komórkę (`Shift + Enter`), aby zapisać zmienną `kurs` w pamięci kernela:

```python
kurs = "Wprowadzenie do data science"
```

### Przykład 3.2 — Odczyt zmiennej w osobnej komórce
W kolejnej komórce zmienna `kurs` jest już znana, mimo że nie ma jej definicji w tej komórce:

```python
print("Aktualny przedmiot:", kurs)
```

### Przykład 3.3 — Pułapka „ukrytego stanu” i nadpisywania zmiennych
Załóżmy, że definiujemy zmienną `liczba_punktow = 100`:

```python
liczba_punktow = 100
print("Wartość początkowa:", liczba_punktow)
```

A w następnej komórce zwiększamy ją o `20`. **Uwaga:** uruchom poniższą komórkę **trzykrotnie** pod rząd za pomocą `Ctrl + Enter` i obserwuj wynik oraz licznik wykonania `[...]`:

```python
liczba_punktow = liczba_punktow + 20
print("Po dodaniu 20:", liczba_punktow)
```

Zauważ, że wielokrotne uruchamianie tej samej komórki modyfikującej zmienną za każdym razem zmienia stan w pamięci kernela (`120` → `140` → `160`), mimo że w komórce wyżej nadal widnieje zapis `liczba_punktow = 100`!

> **Złota zasada higieny pracy w Jupyter Notebooku:**  
> Zanim uznasz swoją analizę lub zadanie za skończone, zawsze użyj na górnym pasku notatnika przycisku **Restart** (czyści pamięć kernela do zera), a następnie **Run All** (wykonuje wszystkie komórki po kolei z góry na dół). Tylko wtedy masz pewność, że Twój notatnik jest odtwarzalny (*reproducible*).

### Przykład 4.1 — Sprawdzenie wersji interpretera Pythona i ścieżki do `.venv`

### Przykład 4.2 — Sprawdzenie informacji o systemie operacyjnym
Moduł `platform` z biblioteki standardowej ([https://docs.python.org/3.14/library/platform.html](https://docs.python.org/3.14/library/platform.html)) pozwala odczytać nazwę systemu operacyjnego oraz implementację Pythona:

```python
import platform

print("System operacyjny:", platform.system(), platform.release())
print("Implementacja    :", platform.python_implementation())
```

## 5. Komendy magiczne IPython (`%who`, `%whos`, `%timeit`) oraz pomoc `?`

### Wprowadzenie
Jądro Jupyter Notebooka opiera się na rozszerzonym interpreterze **IPython** ([https://ipython.readthedocs.io/en/stable/interactive/magics.html](https://ipython.readthedocs.io/en/stable/interactive/magics.html)), który oprócz standardowej składni Pythona udostępnia tzw. **komendy magiczne (*magic commands*)**:
- Komendy liniowe zaczynają się od pojedynczego znaku `%` (np. `%who`, `%whos`, `%timeit`) i działają na pojedynczej linii kodu.
- Komendy komórkowe zaczynają się od podwójnego znaku `%%` (np. `%%timeit`) i obejmują cały kod znajdujący się w danej komórce.
- Znak zapytania `?` postawiony po nazwie obiektu lub funkcji (np. `len?`) wyświetla okienko z dokumentacją (*docstringiem*) tego elementu.

### Przykład 5.1 — Wyświetlenie listy wszystkich zmiennych w pamięci kernela (`%who`)

```python
%who
```

### Przykład 5.2 — Szczegółowa tabela zmiennych wraz z ich typem i wartością (`%whos`)
Komenda `%whos` jest niezwykle pomocna podczas analizy danych, gdy chcemy szybko sprawdzić, jakie zmienne i obiekty mamy aktualnie w pamięci RAM:

```python
%whos
```

### Przykład 5.3 — Pomiar czasu wykonania instrukcji (`%timeit`)
Komenda `%timeit` wielokrotnie wykonuje podaną instrukcję i oblicza średni czas jej wykonania wraz z odchyleniem standardowym:

```python
%timeit sum(range(10_000))
```
