## 1. Operatory arytmetyczne i priorytety wykonywania działań

### Wprowadzenie
Zanim szczegółowo omówimy poszczególne typy danych, poznamy **operatory**, które pozwalają przekształcać wartości liczbowe i logiczne. W analizie danych operatory arytmetyczne służą nie tylko do prostych kalkulacji finansowych, ale również do:
- wyliczania liczby pełnych paczek/stron wyników (dzielenie całkowite `//`),
- sprawdzania podzielności, parzystości lub cyklicznego próbkowania co $n$-tego rekordu (operator reszty z dzielenia `%`),
- skalowania cech i obliczania błędów średniokwadratowych (potęgowanie `**`).

Python wykonuje działania w wyrażeniach zgodnie z matematyczną kolejnością (priorytetami operatorów — [https://docs.python.org/3.14/reference/expressions.html#operator-summary](https://docs.python.org/3.14/reference/expressions.html#operator-summary)):
1. Najpierw nawiasy `(...)`,
2. Następnie potęgowanie `**`,
3. Następnie mnożenie `*`, dzielenie zwykłe `/`, dzielenie całkowite `//` oraz modulo `%` (wykonywane od lewej do prawej),
4. Na końcu dodawanie `+` oraz odejmowanie `-`.

Prześledźmy każdy operator w osobnej komórce kodu.

---

### Przykład 1.1 — Dodawanie (`+`), odejmowanie (`-`) oraz mnożenie (`*`)
Jeżeli oba argumenty są liczbami całkowitymi (`int`), wynik dodawania, odejmowania i mnożenia również pozostaje liczbą całkowitą (`int`):

```python
przychod = 1250
koszt = 800
marza_sztuka = przychod - koszt

print("Marża na sztuce:", marza_sztuka, "| Typ:", type(marza_sztuka))
print("Marża dla 3 sztuk:", marza_sztuka * 3)
```

### Przykład 1.2 — Dzielenie zwykłe (`/`) zawsze zwraca typ zmiennoprzecinkowy (`float`)
W Pythonie 3 operator pojedynczego ukośnika `/` **zawsze zwraca liczbę zmiennoprzecinkową (`float`)**, nawet wtedy, gdy liczby dzielą się przez siebie bez reszty (np. `10 / 2` daje `5.0`, a nie `5`):

```python
print("1 / 2  =", 1 / 2)
print("10 / 2 =", 10 / 2, "| Typ:", type(10 / 2))
```

### Przykład 1.3 — Dzielenie całkowite (`//` — *floor division*)
Operator podwójnego ukośnika `//` zwraca część całkowitą ilorazu (zaokrągloną w dół w kierunku $-\infty$). Jest niezwykle przydatny np. gdy mamy `127` zamówień i pakujemy je po `25` sztuk na jedną stronę raportu:

```python
liczba_rekordow = 127
rozmiar_strony = 25

pelne_strony = liczba_rekordow // rozmiar_strony
print("Liczba pełnych stron po 25 rekordów:", pelne_strony)
```

### Przykład 1.4 — Pułapka dzielenia całkowitego (`//`) dla liczb ujemnych
Ponieważ `//` zaokrągla wynik **w dół (do mniejszej liczby na osi liczbowej)**, dla liczb ujemnych `-7 // 2` daje `-4` (bo `-3.5` zaokrąglone w dół to `-4`, a nie `-3`!):

```python
print(" 7 // 2 =", 7 // 2)
print("-7 // 2 =", -7 // 2)
```

### Przykład 1.5 — Operator modulo (`%`), czyli reszta z dzielenia
Operator `%` zwraca resztę pozostałą po dzieleniu całkowitym. W naszym przykładzie ze stronami raportu (`127` rekordów po `25` na stronę) pozwoli policzyć, ile rekordów zostanie na ostatniej, niepełnej stronie:

```python
pozostale_rekordy = 127 % 25
print("Rekordy na ostatniej niepełnej stronie:", pozostale_rekordy)
```

Operator `%` jest również standardowym sposobem sprawdzania, czy liczba jest parzysta (`x % 2 == 0`) lub podzielna przez inną liczbę:

```python
print("Reszta z dzielenia 14 przez 2:", 14 % 2)
print("Reszta z dzielenia 15 przez 2:", 15 % 2)
```

### Przykład 1.6 — Funkcja wbudowana `divmod(a, b)` — iloraz i reszta jednocześnie
> **Dokumentacja:** [https://docs.python.org/3.14/library/functions.html#divmod](https://docs.python.org/3.14/library/functions.html#divmod)

Gdy potrzebujemy jednocześnie wyniku `a // b` oraz `a % b` (np. zamieniając łączny czas trwania sesji w sekundach na minuty i sekundy), zamiast liczyć oba działania osobno, używamy wbudowanej funkcji `divmod(a, b)`:

```python
czas_sesji_sekundy = 385
minuty, sekundy = divmod(czas_sesji_sekundy, 60)

print(f"Czas sesji ({czas_sesji_sekundy} s) to: {minuty} min {sekundy} s")
```

### Przykład 1.7 — Operator potęgowania (`**`) oraz pierwiastkowanie
Podwójna gwiazdka `**` podnosi liczbę do potęgi. Pamiętając z matematyki, że $\sqrt{x} = x^{0.5}$, tym samym operatorem możemy obliczyć pierwiastek kwadratowy bez importowania dodatkowych modułów:

```python
kwadrat = 5 ** 2
szescian = 5 ** 3
pierwiastek_z_25 = 25 ** 0.5

print("5^2 =", kwadrat, "| 5^3 =", szescian, "| sqrt(25) =", pierwiastek_z_25)
```

### Przykład 1.8 — Kolejność wykonywania działań i rola nawiasów
W poniższym wyrażeniu najpierw wykona się mnożenie i dzielenie (`2 * 3 / 4.0`), a dopiero na końcu dodawanie `1 + ...`. Nawiasy pozwalają jawnie wymusić pożądaną kolejność:

```python
wynik_bez_nawiasow = 1 + 2 * 3 / 4.0
wynik_z_nawiasami = (1 + 2) * 3 / 4.0

print("Bez nawiasów:", wynik_bez_nawiasow)
print("Z nawiasami :", wynik_z_nawiasami)
```

## 2. Operatory przypisania oraz operator przypisania w wyrażeniu (`:=`)

### Wprowadzenie
Pojedynczy znak równości `=` w Pythonie **nie oznacza matematycznej równości**, lecz **przypisanie referencji**: najpierw obliczana jest prawa strona wyrażenia, a następnie lewa strona (nazwa zmiennej) zaczyna wskazywać na wyliczony obiekt.

Python udostępnia kilka bardzo wygodnych rozszerzeń przypisania:
- **Wielokrotne przypisanie w jednej linii:** `a, b = 10, 20` (oraz elegancką zamianę wartości dwóch zmiennych miejscami: `a, b = b, a`).
- **Skrócone operatory przypisania (*augmented assignment*):** `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=` — modyfikują wartość zmiennej w miejscu (np. `suma += kwota` to czytelny skrót od `suma = suma + kwota`).
- **Operator przypisania w wyrażeniu `:=` (*walrus operator* — operator morsa, wprowadzony w [PEP 572](https://peps.python.org/pep-0572/)):** pozwala jednocześnie przypisać wartość do nowej zmiennej **i** od razu użyć tej wartości wewnątrz warunku `if` lub `while`.

---

### Przykład 2.1 — Deklaracja wielu zmiennych w jednej linii i zamiana wartości (`swap`)

```python
min_prog, max_prog = 100, 500
print("Przed zamianą:", min_prog, max_prog)

# Zamiana wartości zmiennych miejscami bez zmiennej pomocniczej:
min_prog, max_prog = max_prog, min_prog
print("Po zamianie   :", min_prog, max_prog)
```

### Przykład 2.2 — Skrócone operatory przypisania (`+=`, `*=`)

```python
suma_koszyka = 200.0
suma_koszyka += 50.0   # dodanie produktu za 50 PLN (odpowiednik: suma_koszyka = suma_koszyka + 50.0)
print("Po dodaniu produktu :", suma_koszyka)

suma_koszyka *= 0.90   # naliczenie 10% rabatu (pomnożenie przez 0.90)
print("Po naliczeniu rabatu:", suma_koszyka)
```

### Przykład 2.3 — Operator przypisania w wyrażeniu `:=` (*walrus operator*)
Wyobraź sobie, że sprawdzasz, czy lista pomiarów przekracza dopuszczalny limit `5` elementów, a jeśli tak — chcesz w komunikacie wypisać jej dokładną długość. Zamiast wywoływać `len(pomiary)` dwa razy lub tworzyć zmienną przed `if`, możemy użyć operatora `:=`:

```python
pomiary = [12.1, 14.5, 11.8, 15.0, 13.2, 16.4, 14.9]

if (liczba_pomiarow := len(pomiary)) > 5:
    print(f"Zbiór jest za duży ({liczba_pomiarow} elementów, oczekiwano maksymalnie 5).")
```

## 3. Operatory porównania, typ `bool`, operatory logiczne oraz wartości *falsy*

> **Dokumentacja:** [https://docs.python.org/3.14/library/stdtypes.html#truth-value-testing](https://docs.python.org/3.14/library/stdtypes.html#truth-value-testing) oraz [https://docs.python.org/3.14/library/stdtypes.html#comparisons](https://docs.python.org/3.14/library/stdtypes.html#comparisons)

### Wprowadzenie
Operatory porównania (`>`, `<`, `>=`, `<=`, `==`, `!=`) zawsze zwracają wartość logiczną typu **`bool`**, czyli `True` (prawda) lub `False` (fałsz).

W Pythonie warto znać trzy wyjątkowo praktyczne cechy logiki:
1. **Porównania łańcuchowe (*chained comparisons*):** zapis matematyczny $0 \le x \le 100$ możemy w Pythonie zapisać dosłownie jako `0 <= x <= 100` (jest to równoważne `(0 <= x) and (x <= 100)`, przy czym `x` jest obliczane tylko raz!).
2. **Wartości traktowane jako fałsz (*Falsy values*):** w warunkach `if` oraz w konwersji `bool(obiekt)` jako `False` traktowane są dokładnie następujące wartości:
   - `False` oraz `None`,
   - zero dowolnego typu liczbowego: `0`, `0.0`, `0j`, `Decimal(0)`,
   - puste sekwencje i kolekcje: pusty napis `""`, pusta lista `[]`, pusta krotka `()`, pusty słownik `{}`, pusty zbiór `set()`, pusty `range(0)`,
   - obiekty klas definiujących metodę `__bool__()` zwracającą `False` (lub `__len__()` zwracającą `0`).
   **Wszystkie inne wartości w Pythonie są traktowane jako prawda (*Truthy*)!**
3. **Leniwe wartościowanie (*Short-circuit evaluation*) operatorów `and` oraz `or`:**
   - `x or domyslna` zwraca `x`, jeśli `x` jest *truthy*, a w przeciwnym razie zwraca `domyslna`. To bardzo popularny idiom do podstawiania wartości domyślnej, gdy pole tekstowe z formularza jest puste (`""`) lub równe `None`!

---

### Przykład 3.1 — Operatory porównania (`==`, `!=`, `<`, `>`, `<=`, `>=`)

```python
cena_a = 150
cena_b = 200

print("cena_a < cena_b :", cena_a < cena_b)
print("cena_a == cena_b:", cena_a == cena_b)
print("cena_a != cena_b:", cena_a != cena_b)
```

### Przykład 3.2 — Porównania łańcuchowe (`dolny_prog <= wartosc <= gorny_prog`)
Sprawdźmy, czy rabat procentowy mieści się w dozwolonym przedziale od `0` do `50` procent:

```python
rabat_proc = 25
czy_poprawny_rabat = 0 <= rabat_proc <= 50

print("Czy rabat 25% jest w przedziale [0, 50]?", czy_poprawny_rabat)
```

### Przykład 3.3 — Operatory logiczne `and` (koniunkcja), `or` (alternatywa), `not` (negacja)

```python
ma_konto_aktywne = True
ma_zaleglosci = False

print("Może złożyć zamówienie (and + not):", ma_konto_aktywne and not ma_zaleglosci)
print("Wymaga kontaktu (or)               :", not ma_konto_aktywne or ma_zaleglosci)
```

### Przykład 3.4 — Sprawdzanie prawdziwości (*Truth Value Testing*) konstruktorem `bool()`
Przetestujmy, jak funkcja `bool()` ocenia puste i niepuste wartości różnego typu:

```python
print("bool(0)      ->", bool(0),      "| bool(42)       ->", bool(42))
print("bool('')     ->", bool(""),     "| bool('Olsztyn')->", bool("Olsztyn"))
print("bool([])     ->", bool([]),     "| bool([1, 2])   ->", bool([1, 2]))
print("bool(None)   ->", bool(None))
```

### Przykład 3.5 — Praktyczny idiom z operatorem `or` do zastępowania brakujących napisów
Gdy wczytujemy dane z formularza lub pliku, puste pole `""` (lub `None`) jest *falsy*. Dzięki temu zapis `wczytane_miasto or "Nieznane"` automatycznie podstawi `"Nieznane"`, gdy pole było puste:

```python
wczytane_miasto_1 = ""
wczytane_miasto_2 = "Gdańsk"

miasto_1 = wczytane_miasto_1 or "Nieznane"
miasto_2 = wczytane_miasto_2 or "Nieznane"

print("Miasto 1:", miasto_1, "| Miasto 2:", miasto_2)
```

## 4. Operatory tożsamości (`is`), przynależności (`in`) oraz operatory bitowe

> **Dokumentacja:** [https://docs.python.org/3.14/reference/expressions.html#is-not](https://docs.python.org/3.14/reference/expressions.html#is-not) oraz [https://docs.python.org/3.14/library/stdtypes.html#bitwise-operations-on-integer-types](https://docs.python.org/3.14/library/stdtypes.html#bitwise-operations-on-integer-types)

### Wprowadzenie
1. **Operatory tożsamości (`is` oraz `is not`):** sprawdzają, czy dwie zmienne wskazują na **dokładnie ten sam obiekt w pamięci komputera** (`id(a) == id(b)`), podczas gdy operator `==` sprawdza jedynie, czy obiekty mają **równą wartość**.
   > **Ważna reguła PEP 8:** Do sprawdzania, czy zmienna ma wartość `None`, zawsze używamy `zmienna is None` lub `zmienna is not None` (ponieważ w całym programie istnieje tylko jeden unikalny obiekt `None` — tzw. *singleton*).
2. **Operatory przynależności (`in` oraz `not in`):** sprawdzają, czy element występuje wewnątrz kolekcji (listy, zbioru, słownika) lub czy podciąg znaków znajduje się wewnątrz dłuższego napisu.
3. **Operatory bitowe (`&`, `|`, `^`, `~`, `<<`, `>>`):** wykonują operacje na pojedynczych bitach liczb całkowitych. Dlaczego warto je kojarzyć w Data Science? W bibliotece **`pandas`** operatory `&` (AND), `|` (OR) oraz `~` (NOT) zostały przeciążone do łączenia warunków logicznych dla całych kolumn tabeli!

---

### Przykład 4.1 — Różnica pomiędzy `==` (równość wartości) a `is` (tożsamość obiektu w pamięci)
Dwie osobno utworzone listy o tych samych elementach mają równą zawartość (`==` daje `True`), ale leżą pod dwoma różnymi adresami w pamięci (`is` daje `False`):

```python
koszyk_1 = ["Mysz", "Klawiatura"]
koszyk_2 = ["Mysz", "Klawiatura"]

print("Czy mają tę samą zawartość (==)?", koszyk_1 == koszyk_2)
print("Czy to ten sam obiekt w pamięci (is)?", koszyk_1 is koszyk_2)
```

### Przykład 4.2 — Poprawne sprawdzanie braku wartości za pomocą `is None`

```python
ocena_klienta = None

print("Czy brak oceny?", ocena_klienta is None)
print("Czy ocena została wystawiona?", ocena_klienta is not None)
```

### Przykład 4.3 — Operatory przynależności (`in`, `not in`) na napisach i listach

```python
dozwolone_waluty = ["PLN", "EUR", "USD"]
wybrana_waluta = "PLN"
opis_transakcji = "Opłata za zamówienie ZAM-2026-105"

print("Czy PLN jest na liście walut?", wybrana_waluta in dozwolone_waluty)
print("Czy 'CHF' nie jest obsługiwany?", "CHF" not in dozwolone_waluty)
print("Czy opis dotyczy roku 2026?", "2026" in opis_transakcji)
```

### Przykład 4.4 — Operatory bitowe (`&`, `|`, `^`, `<<`, `>>`)
Na liczbach całkowitych operatory te działają na reprezentacji dwójkowej (binarnej), np. `5` to binarnie `0b0101`, a `3` to `0b0011`:

```python
x = 0b0101  # 5 dziesiętnie
y = 0b0011  # 3 dziesiętnie

print(f"x & y (AND) : {x & y} (binarnie: {x & y:04b})")
print(f"x | y (OR)  : {x | y} (binarnie: {x | y:04b})")
print(f"x ^ y (XOR) : {x ^ y} (binarnie: {x ^ y:04b})")
print(f"1 << 3      : {1 << 3} (przesunięcie bitowe w lewo = mnożenie przez 2^3)")
```

## 5. Typy liczbowe (`int`, `float`), konwersje oraz dokładność obliczeń (`Decimal`)

> **Dokumentacja:**
> - Liczby zmiennoprzecinkowe i ich ograniczenia: [https://docs.python.org/3.14/tutorial/floatingpoint.html](https://docs.python.org/3.14/tutorial/floatingpoint.html)
> - Moduł `decimal`: [https://docs.python.org/3.14/library/decimal.html](https://docs.python.org/3.14/library/decimal.html)

### Wprowadzenie
W Pythonie istnieją trzy wbudowane typy liczbowe:
1. **`int`** — liczby całkowite o **nieograniczonej precyzji** (w Pythonie 3 nie ma problemu przepełnienia 32- czy 64-bitowego; wielkość liczby całkowitej ogranicza jedynie dostępna pamięć RAM). Dla poprawy czytelności dużych liczb w kodzie możemy rozdzielać tysiące znakiem podkreślenia `_`, np. `1_000_000` ([PEP 515](https://peps.python.org/pep-0515/)).
2. **`float`** — liczby zmiennoprzecinkowe podwójnej precyzji (64-bitowy standard IEEE 754).
3. **`complex`** — liczby zespolone (np. `3 + 4j`), używane głównie w przetwarzaniu sygnałów i elektrotechnice.

#### Dlaczego w komputerze `0.1 + 0.2` nie jest idealnie równe `0.3`?
Komputery przechowują liczby typu `float` jako ułamki **dwójkowe (binarne)**, a nie dziesiętne. Podobnie jak ułamka
$\frac{1}{3}$ nie da się zapisać skończoną liczbą cyfr w systemie dziesiętnym ($0.33333...$), tak samo dziesiętnego ułamka $0.1$ (
$\frac{1}{10}$) nie da się zapisać skończonym rozwinięciem w systemie dwójkowym ($0.0001100110011..._2$). W pamięci zapisywane jest więc niezwykle dokładne, 53-bitowe przybliżenie.

Sprawdźmy ten efekt i poznajmy trzy sposoby radzenia sobie z nim w praktyce!

### Przykład 5.5 — Rozwiązanie 2 (dla obliczeń finansowych i księgowych): `decimal.Decimal`
W systemach finansowych (ceny, podatki, prowizje) nawet ułamki grosza wynikające z reprezentacji binarnej są niedopuszczalne. Używamy wtedy klasy `Decimal` z wbudowanego modułu `decimal`.

> **Kluczowa zasada przy tworzeniu obiektów `Decimal`:**  
> Wartość ułamkową zawsze przekazujemy do konstruktora `Decimal` jako **łańcuch znaków (`str`)**, np. `Decimal('0.1')`! Jeśli przekażesz tam `float` (`Decimal(0.1)`), przekażesz liczbę już obarczoną niedokładnością binarną.

```python
from decimal import Decimal

# ŹLE (przekazanie niedokładnego float do Decimal):
print("Decimal z float:", (Decimal(0.1) + Decimal(0.2)) == Decimal(0.3))

# DOBRZE (przekazanie dokładnego zapisu tekstowego str do Decimal):
kwota_a = Decimal("0.1")
kwota_b = Decimal("0.2")
suma_dec = kwota_a + kwota_b

print("Decimal ze str :", suma_dec, "| Czy równe Decimal('0.3')?", suma_dec == Decimal("0.3"))
```
