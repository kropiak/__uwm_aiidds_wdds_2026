## 1. Typ `str` w Data Science — deklaracja, indeksowanie, wycinanie (*slicing*) i niemutowalność

### Wprowadzenie
Typ łańcuchowy **`str`** (ciąg znaków Unicode) jest najbardziej powszechnym typem danych w codziennej pracy analityka danych. Dane wczytywane z plików CSV, logów serwerowych, formularzy webowych czy odpowiedzi API początkowo docierają do programu jako surowy tekst, który dopiero na etapie przygotowania i czyszczenia danych (*data cleaning* / *data wrangling*) oczyszczamy, dzielimy na części i konwertujemy na typy liczbowe lub daty.

Warto pamiętać o trzech fundamentalnych właściwościach klasy `str` w Pythonie:
1. **Konstruktor `str(obiekt)` i metoda `__str__()`:** każdy obiekt w Pythonie dziedziczy po klasie bazowej `object`, która dostarcza metodę specjalną `__str__()`. Wywołanie `str(5)` lub `print(5)` uruchamia pod spodem właśnie metodę `(5).__str__()`, zamieniającą obiekt na postać tekstową.
2. **Sekwencja znaków (indeksowanie i *slicing*):** łańcuch tekstowy to uporządkowana sekwencja znaków. Do pojedynczego znaku odwołujemy się przez indeks w nawiasach kwadratowych `[i]` (od `0` od lewej lub od `-1` od prawej), a podciąg wycinamy składnią `[start:stop:step]` (indeks `stop` **nie wchodzi** do wycinka!).
3. **Niemutowalność (*Immutability*):** po utworzeniu obiektu `str` w pamięci **nie można zmienić żadnego pojedynczego znaku wewnątrz niego**. Wszystkie metody klasy `str` (np. `.lower()`, `.replace()`) zawsze zwracają **nowy obiekt `str`**, pozostawiając oryginał bez zmian.

Przećwiczmy te właściwości krok po kroku na pojedynczych komórkach kodu.

### Przykład 1.1 — Sposoby deklaracji `str` oraz działanie konstruktora `str()` i `__str__()`

````python
miasto = "Olsztyn"
cytat = 'Raport kwartalny "Q4 2026"'
liczba_rekordow = 1000
liczba_jako_tekst = str(liczba_rekordow)

print(miasto, "|", cytat)
print("str(1000):", liczba_jako_tekst, "| Typ:", type(liczba_jako_tekst))
print("Wywołanie jawne (1000).__str__():", (1000).__str__())
````

### Przykład 1.2 — Indeksowanie dodatnie (od początku) oraz ujemne (od końca ciągu)
Załóżmy, że analizujemy ustandaryzowany kod zamówienia `"PL-2026-OLS-0042"`:
- indeks `0` to pierwszy znak (`'P'`),
- indeks `-1` to ostatni znak (`'2'`), a `-2` to przedostatni znak (`'4'`).

````python
kod_zamowienia = "PL-2026-OLS-0042"

print("Pierwszy znak (indeks 0)   :", kod_zamowienia[0])
print("Drugi znak (indeks 1)      :", kod_zamowienia[1])
print("Ostatni znak (indeks -1)   :", kod_zamowienia[-1])
print("Przedostatni znak (-2)     :", kod_zamowienia[-2])
print("Całkowita długość len()    :", len(kod_zamowienia))
````

### Przykład 1.3 — Wycinanie podciągów (*slicing*) `[start:stop]`
Wyciągnijmy z kodu `"PL-2026-OLS-0042"` poszczególne pola informacyjne (pamiętaj, że znak na pozycji `stop` nie jest włączany do wyniku):

````python
kraj = kod_zamowienia[:2]         # od początku do indeksu 2 (czyli indeksy 0 i 1)
rok = kod_zamowienia[3:7]         # od indeksu 3 do 7 (czyli indeksy 3, 4, 5, 6)
oddzial = kod_zamowienia[8:11]    # od indeksu 8 do 11
numer_porzadkowy = kod_zamowienia[12:]  # od indeksu 12 do samego końca

print("Kraj:", kraj, "| Rok:", rok, "| Oddział:", oddzial, "| Nr:", numer_porzadkowy)
````

### Przykład 1.4 — Trzeci parametr wycinka `[start:stop:step]` oraz odwracanie napisu `[::-1]`
Parametr `step` określa krok, z jakim pobierane są znaki. Krok ujemny (`-1`) porusza się od końca do początku, co pozwala w jednej instrukcji odwrócić dowolny ciąg znaków:

````python
nazwisko = "Kowalski"

print("Co drugi znak [::2]          :", nazwisko[::2])
print("3 ostatnie znaki od końca    :", nazwisko[:-4:-1])
print("Odwrócony cały napis [::-1]  :", nazwisko[::-1])
````

### Przykład 1.5 — Łączenie (`+`) oraz powielanie (`*`) łańcuchów znaków
Operator `+` skleja (konkatenuje) napisy, natomiast `*` powtarza napis zadaną liczbę razy (przydatne m.in. przy rysowaniu linii separatorów w raportach konsolowych):

````python
pelna_nazwa = "Jan" + " " + "Kowalski"
linia_raportu = "=" * 35

print(linia_raportu)
print("Klient:", pelna_nazwa)
print(linia_raportu)
````

### Przykład 1.6 — Dowód niemutowalności typu `str`
Próba podmiany pojedynczego znaku przez `nazwisko[0] = "P"` zakończy się błędem `TypeError: 'str' object does not support item assignment`. Co więcej, wywołanie metody takiej jak `.lower()` nie zmienia zmiennej `nazwisko`, dopóki nie przypiszemy zwróconego wyniku z powrotem do zmiennej:

````python
nazwisko = "Kowalski"
print("Wynik nazwisko.lower():", nazwisko.lower())
print("Oryginalna zmienna    :", nazwisko)  # nadal "Kowalski"!

# Aby utworzyć napis ze zmienioną pierwszą literą, budujemy nowy obiekt str:
nowe_nazwisko = "P" + nazwisko[1:]
print("Nowy obiekt str       :", nowe_nazwisko)
````

### Grupa 2.1 — Oczyszczanie brzegów napisu: `.strip()`, `.lstrip()`, `.rstrip()` oraz nowoczesne `.removeprefix()` i `.removesuffix()`
- `.strip()` usuwa białe znaki (spacje, tabulatory `	`, znaki nowej linii `
`) z początku i końca napisu (`.lstrip()` tylko z lewej, `.rstrip()` tylko z prawej).
- **Ważna pułapka i nowość (Python 3.9+):** jeśli do `.strip("abc")` przekażesz argument tekstowy, Python potraktuje go jako **zbiór pojedynczych znaków do usunięcia**, a nie jako cały wyraz! Dlatego do bezpiecznego odcinania znanego przedrostka lub przyrostka (np. `"ID_"`, `".csv"`, `" PLN"`) zawsze używamy metod **`.removeprefix()`** oraz **`.removesuffix()`**.

````python
surowa_nazwa = "   \t Olsztyn \n  "
print("Po .strip():", repr(surowa_nazwa.strip()))
````

````python
# Porównanie pułapki .rstrip() z bezpieczną metodą .removesuffix():
nazwa_pliku = "raport_marzec.csv"

print("Błędne .rstrip('.csv')      :", nazwa_pliku.rstrip(".csv"))        # ucięło też 'c' z słowa 'marzec'!
print("Poprawne .removesuffix('.csv'):", nazwa_pliku.removesuffix(".csv"))  # usuwa dokładnie sufiks '.csv'
print("Poprawne .removeprefix()     :", "ID_98412".removeprefix("ID_"))
````

### Grupa 2.2 — Podział i łączenie tekstu: `.split()`, `.partition()`, `.splitlines()` oraz `.join()`
> **Ważny szczegół z dokumentacji dotyczący `.split()`:**  
> Wywołanie `tekst.split(' ')` (ze spacją jako argumentem) dzieli napis dokładnie na każdej pojedynczej spacji — jeśli w tekście są dwie spacje obok siebie, w wynikowej liście pojawią się puste napisy `""`.  
> Natomiast wywołanie `tekst.split()` **bez argumentu** (lub z `None`) stosuje inteligentny algorytm: traktuje dowolny ciąg sąsiadujących białych znaków (wiele spacji, tabulatory, nowe linie) jako **jeden separator** i automatycznie ignoruje spacje na brzegach!

````python
zdanie_z_wieloma_spacjami = "  Python   3.14   w   Data  Science  "

print("split(' ') :", zdanie_z_wieloma_spacjami.split(" "))
print("split()    :", zdanie_z_wieloma_spacjami.split())
````

Gdy dzielimy wiersz pliku CSV rozdzielany średnikiem `;` i chcemy następnie złączyć elementy nowym separatorem (np. przecinkiem `,`), łączymy `.split()` z metodą `.join()`:

````python
wiersz_csv = "Warszawa;Elektronika;299.90;BLIK"
pola = wiersz_csv.split(";")
nowy_wiersz = " | ".join(pola)

print("Lista pól :", pola)
print("Po .join():", nowy_wiersz)
````

Metoda **`.partition(sep)`** dzieli napis dokładnie w miejscu **pierwszego** wystąpienia separatora na 3-elementową krotkę `(przed, separator, po)`. Jest niezastąpiona przy parsowaniu par `klucz: wartość` lub `login@domena`, gdzie sam tekst po prawej stronie może zawierać kolejne dwukropki:

````python
wpis_logu = "ERROR:2026-11-17:Błąd połączenia z bazą: timeout"
poziom, sep, reszta_komunikatu = wpis_logu.partition(":")

print("Poziom:", poziom, "| Reszta:", reszta_komunikatu)
````

Metody `.startswith()` oraz `.endswith()` potrafią przyjąć jako argument **krotkę wielu szukanych przedrostków/przyrostków** (np. sprawdzenie jednym wywołaniem, czy plik ma rozszerzenie `.csv`, `.tsv` lub `.xlsx`!).

### Grupa 2.5 — Walidacja zawartości tekstu (metody predykatowe `is...`)
Zanim wywołamy `int(tekst)` lub `float(tekst)` na niepewnych danych, możemy sprawdzić z jakich znaków składa się napis:

````python
probka_1 = "12345"
probka_2 = "Python3"
probka_3 = "Olsztyn"
probka_float = "129.99"

print("'12345'.isdecimal() :", probka_1.isdecimal())
print("'Python3'.isalnum() :", probka_2.isalnum(), "| isalpha():", probka_2.isalpha())
print("'Olsztyn'.istitle() :", probka_3.istitle(), "| isascii():", probka_3.isascii())

# Sprytny sposób sprawdzenia czy napis reprezentuje dodatnią liczbę float (z jedną kropką):
czy_dodatni_float = probka_float.replace(".", "", 1).isdecimal()
print("Czy '129.99' to poprawny zapis dziesiętny?", czy_dodatni_float)
````

## 3. Formatowanie łańcuchów znaków: nowoczesne `f-stringi` (PEP 498 / PEP 701) oraz `t-stringi` (PEP 750)

> **Dokumentacja:**
> - Składnia formatowania napisów: [https://docs.python.org/3.14/library/string.html#format-string-syntax](https://docs.python.org/3.14/library/string.html#format-string-syntax)
> - Przewodnik Real Python po f-stringach: [https://realpython.com/python-f-strings/](https://realpython.com/python-f-strings/)

### Wprowadzenie
Formatowanie łańcuchów polega na łączeniu szablonu tekstowego ze zmiennymi i określaniu sposobu ich prezentacji (liczby miejsc po przecinku, wyrównania w kolumnach, znaków walut czy procentów).

W historii Pythona funkcjonowały kolejno trzy podejścia:
1. **Operator `%` (z czasów Pythona 2):** np. `'Wersja %.1f' % 3.14` — dziś spotykany już tylko w bardzo starym kodzie i konfiguracji modułu `logging`.
2. **Metoda `.format()` (Python 2.6 / 3.0):** np. `'Lubię {1} oraz {0}'.format('R', 'Pythona')`.
3. **f-stringi (`f"..."` — od Pythona 3.6, znacznie rozszerzone w Pythonie 3.12 przez [PEP 701](https://peps.python.org/pep-0701/)):** obecny, najszybszy i najczytelniejszy standard w Pythonie. Ponadto w **Pythonie 3.14** pojawił się nowy kuzyn f-stringów: **napisy szablonowe `t"..."` (*template strings*, [PEP 750](https://peps.python.org/pep-0750/))**.

Zobaczmy najpierw w jednej komórce historyczny zapis `%` oraz `.format()` (abyś potrafił go rozpoznać w cudzym kodzie), a następnie przećwiczmy pełne możliwości nowoczesnych **f-stringów** w osobnych komórkach!

---

### Przykład 3.1 — Kontekst historyczny: operator `%` oraz metoda `.format()`

````python
# Starszy zapis z operatorem % oraz metodą .format():
print("Operator %%     : Język %s w wersji %.2f (szesnastkowo: %X)" % ("Python", 3.14, 14))
print("Metoda .format(): Język {0} ma już ponad {wiek} lat.".format("Python", wiek=35))
````

### Przykład 3.2 — Podstawy `f-stringów` i wykonywanie wyrażeń wewnątrz `{...}`
Wewnątrz nawiasów klamrowych `{...}` w f-stringu możemy umieścić dowolne wyrażenie Pythona (działanie matematyczne, wywołanie metody), a po dwukropku `:` określić specyfikator formatu (np. `.2f` oznacza liczbę zmiennoprzecinkową z dokładnością do 2 miejsc po przecinku):

````python
imie = "marek"
kwota_1 = 33.33
kwota_2 = 66.67

print(f"Witaj {imie.capitalize()}! Suma wynosi: {kwota_1 + kwota_2:.2f} PLN")
````

### Przykład 3.3 — Formatowanie raportowe: separatory tysięcy (`,`, `_`), procenty (`%`) oraz znak liczby (`+`)
W raportach Data Science czytelność dużych liczb i wskaźników dynamiki ma kluczowe znaczenie:

````python
przychod_roczny = 4_850_120.5
udzial_rynku = 0.2368
zmiana_rr = 0.084

print(f"Przychód (separator _): {przychod_roczny:_.2f} PLN")
print(f"Przychód (separator ,): {przychod_roczny:,.2f} PLN")
print(f"Udział w rynku        : {udzial_rynku:.1%}")
print(f"Dynamika r/r (ze znakiem +): {zmiana_rr:+.2%}")
````

### Przykład 3.4 — Wyrównywanie kolumn w tabelach tekstowych (`<`, `>`, `^`) i znak wypełnienia
Specyfikatory `<N` (do lewej na szerokości `N` znaków), `>N` (do prawej) oraz `^N` (do środka) pozwalają budować idealnie równe tabele w konsoli i logach:

````python
print(f"{'Miasto':<12} | {'Zamówienia':>10} | {'Udział':>8}")
print("-" * 36)
print(f"{'Warszawa':<12} | {1420:>10_d} | {0.412:>8.1%}")
print(f"{'Olsztyn':<12} | {385:>10_d} | {0.112:>8.1%}")
````

### Przykład 3.5 — Szybkie debugowanie za pomocą znaku `=` wewnątrz f-stringa (`f"{wyrazenie = }"`)
Dodanie znaku `=` na końcu wyrażenia wewnątrz `{...}` sprawia, że Python automatycznie wypisuje zarówno **nazwę zmiennej / treść wyrażenia**, jak i **jego obliczoną wartość**:

````python
cena_netto = 200.0
stawka_vat = 0.23

print(f"{cena_netto = } | {stawka_vat = } | {cena_netto * (1 + stawka_vat) = :.2f}")
````

### Przykład 3.6 — Nowość z Pythona 3.12+ (PEP 701): te same cudzysłowy wewnątrz f-stringów
W starszych wersjach Pythona wewnątrz `f"..."` nie wolno było użyć podwójnego cudzysłowu `"` (np. przy odwołaniu do klucza słownika). Od Pythona 3.12 (PEP 701) jest to w pełni dozwolone, podobnie jak zagnieżdżanie f-stringów:

````python
rekord = {"miasto": "  kraków ", "sprzedaz": 1850.5}

print(f"Oddział: {rekord["miasto"].strip().title()} -> {rekord["sprzedaz"]:.2f} PLN")
````

## 4. Sekwencje sterujące (*Escape sequences*), napisy surowe (`r"..."`) oraz znaki Unicode

> **Dokumentacja:** [https://docs.python.org/3.14/reference/lexical_analysis.html#escape-sequences](https://docs.python.org/3.14/reference/lexical_analysis.html#escape-sequences)

### Wprowadzenie
Wewnątrz zwykłych łańcuchów tekstowych znak odwrotnego ukośnika `\` (*backslash*) pełni rolę **znaku ucieczki (*escape character*)**, rozpoczynającego sekwencję sterującą:
- `\n` — przejście do nowej linii (*linefeed*),
- `\t` — tabulator poziomy,
- `\r` — powrót karetki na początek bieżącego wiersza (*carriage return*),
- `\\` — dosłowny znak pojedynczego ukośnika `\`.

Powoduje to klasyczną pułapkę na systemie Windows przy zapisywaniu ścieżek do plików, np. `'\sciezka\rok\numer\task'`, gdzie `\r`, `\n` oraz `\t` zostają zinterpretowane jako znaki sterujące! Aby wyłączyć interpretację znaków ucieczki, poprzedzamy łańcuch literą **`r`** (*raw string* — napis surowy): `r'\sciezka\rok\numer\task'`.

---

### Przykład 4.1 — Pułapka ścieżki ze znakami `\r`, `\n`, `\t` vs napis surowy `r"..."`
Porównajmy wynik wydruku zwykłego napisu oraz napisu surowego `r'...'`:

````python
print("1. Zwykły napis (uszkodzony przez \r, \n, \t):")
print('\sciezka\rok\numer\task')

print("\n2. Napis surowy r'...' (poprawny):")
print(r'\sciezka\rok\numer\task')

print("\n3. Podwójne ukośniki '\\':")
print('\\sciezka\\rok\\numer\\task')
````

### Przykład 4.2 — Znaki z tablicy Unicode (`\u`, `\U`, `\N{...}`) oraz kodowanie UTF-8
W Pythonie 3 każdy znak w napisie `str` jest punktem kodowym standardu **Unicode** ([https://symbl.cc/en/unicode/table/](https://symbl.cc/en/unicode/table/)). Możemy wstawiać znaki specjalne przez:
- `\uXXXX` (4-znakowy kod szesnastkowy),
- `\UXXXXXXXX` (8-znakowy kod szesnastkowy, np. dla emoji),
- `\N{NAZWA ZNAKU UNICODE}` (oficjalna angielska nazwa znaku),
- lub sprawdzić kod liczbowy znaku funkcją `ord()`, a znak dla kodu funkcją `chr()`.

````python
print(r"Symbole przez kod \u i \U :", "\u2764", "\u03b1", "\u20ac", "\u20bf", "\U0001F4CA")
print(r"Symbole przez nazwę \N{}  :", "\N{GREEK SMALL LETTER PI}", "\N{EURO SIGN}", "\N{BITCOIN SIGN}")
print("Kod Unicode znaku 'ą'     :", ord("ą"), "| Znak o kodzie 261:", chr(261))
````
